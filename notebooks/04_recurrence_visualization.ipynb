# 04 — Interactive Recurrence Visualization

**Purpose:** Generate a linked-brushing Bokeh widget that lets you explore recurrence structure in a single trial.

The output is a standalone HTML file with two linked panels:

- **Left panel:** z-scored gaze angle vs time.
- **Right panel:** recurrence plot (scatter of recurrent point pairs).

Click any point in either panel to highlight all of its recurrent partners in both views.

## Requirements

This notebook requires **Bokeh** (`pip install bokeh`).

```python
import sys, os
sys.path.insert(0, os.path.abspath(".."))

import numpy as np
import pandas as pd
```

## Configuration

Set `DEMO = True` to run with the small sample data shipped with this repo.  
Set `DEMO = False` (the default) to use your own segmented trial CSVs.


```python
# ── CONFIG ──────────────────────────────────────────────────────────
# If you are using this repo for the first time, set DEMO = True to
# run the pipeline on the included sample data.
DEMO = False

if DEMO:
    TRIALS_DIR = "../demo/walk_segmented_csvs"
    OUTPUT_HTML = "../demo/output/recurrence_widget.html"
else:
    TRIALS_DIR = "../walk_segmented_csvs"
    OUTPUT_HTML = "../output/recurrence_widget.html"

# Recurrence radius in z-score units
EPSILON = 0.07
```

## Select a trial file

```python
from ipywidgets import Dropdown
from IPython.display import display

csv_files = sorted(
    f for f in os.listdir(TRIALS_DIR) if f.lower().endswith(".csv")
) if os.path.isdir(TRIALS_DIR) else []

if not csv_files:
    print(f"No CSVs found in '{TRIALS_DIR}'. Run Notebook 02 first.")
else:
    file_dropdown = Dropdown(
        options=csv_files,
        value=csv_files[0],
        description="Trial:",
        style={"description_width": "initial"},
        layout={"width": "500px"},
    )
    display(file_dropdown)
```

## Build the interactive widget

The cell below imports the Bokeh-based builder from `neon_gaze.visualization` and generates a standalone HTML file.

```python
from neon_gaze.visualization import build_linked_recurrence_widget

selected_file = file_dropdown.value
csv_path = os.path.join(TRIALS_DIR, selected_file)

os.makedirs(os.path.dirname(OUTPUT_HTML) or ".", exist_ok=True)

layout = build_linked_recurrence_widget(
    csv_path=csv_path,
    epsilon=EPSILON,
    output_html=OUTPUT_HTML,
    show=False,
)

print(f"Saved interactive widget to: {OUTPUT_HTML}")
print(f"Open the HTML file in a browser to explore.")
```
