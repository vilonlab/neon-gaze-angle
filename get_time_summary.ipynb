```python
import os
import glob
import pandas as pd

# Directory containing your CSV files
BASE_DIR = "walk_segmented_csvs"  # change if needed

# Find all CSV files in the directory
csv_paths = sorted(glob.glob(os.path.join(BASE_DIR, "*.csv")))

rows = []

for path in csv_paths:
    fname = os.path.basename(path)
    
    try:
        df = pd.read_csv(path)
    except Exception as e:
        print(f"Could not read {fname}: {e}")
        continue

    if "time_sec" not in df.columns:
        print(f"File {fname} missing 'time_sec' column. Skipping.")
        continue

    # Drop NaNs in time_sec just in case
    ts = df["time_sec"].dropna()
    if ts.empty:
        print(f"File {fname} has no valid 'time_sec' values. Skipping.")
        continue

    rows.append({
        "file name": fname,
        "time_sec[0]": ts.iloc[0],
        "time_sec[end]": ts.iloc[-1],
    })

# Build summary DataFrame
summary_df = pd.DataFrame(rows)

# Save to CSV
output_path = "time_sec_summary.csv"
summary_df.to_csv(output_path, index=False)

summary_df
```
